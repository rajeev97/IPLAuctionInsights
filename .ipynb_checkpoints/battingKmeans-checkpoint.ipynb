```python
# Importing all library
import numpy as np
import pandas as pd
from sklearn.preprocessing import MinMaxScaler
from sklearn.decomposition import PCA
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt
%matplotlib inline
import seaborn as sns
import plotly.express as px
```

```python
#Setting colors for plot
custom_colors =["blue", "sienna", "limegreen", "red", "black",'cyan']
#setting relative path for data
path = "./data"
```

```python
#Reading data
batting_data =pd.read_csv(path+"/salariesFinal.csv")
columns = batting_data.columns.tolist()
cols_to_use = columns[1:len(columns)-1] 
print(columns)
print(cols_to_use)
```

```python
# For batting data only selecting batsmen who satisfy following criteria
# 1) Total score more than 200
# 2) Average score more than 10
# 3) Strike Rate more than 95
data=batting_data[["Player","location","SR", "Runs","Average","4s","6s","50s"]].groupby(["Player","location"],as_index=False).agg({'SR':'mean','Runs':'sum','Average':'mean','4s':'sum', '6s':'sum','50s':'sum'})
data=data[(data['Runs'] > 200) & (data['Average'] > 10) & (data['SR'] > 95) ]
print(len(data))
display(data)
```

```python
#Scaling the data
scaler = MinMaxScaler(feature_range=(0,1))
```

```python
bat_data=data.iloc[:,2:].to_numpy()
#bat_data
```

```python
batting_data_normalized = scaler.fit_transform(bat_data)
```

```python
#print(batting_data_normalized)
```

```python
# PCA , Reducing to two variables
pca = PCA(n_components=2)
pca.fit(batting_data_normalized)
reduced_rep_batting_data = pca.transform(batting_data_normalized)
```

```python
reduced_rep_batting_data
```

```python
#K-Mean: Getting The Optimal Number Of Clusters
sumsqdis = []
K = range(1,10)
for num_clusters in K :
    kmeans = KMeans(n_clusters=num_clusters)
    kmeans.fit(reduced_rep_batting_data)
    sumsqdis.append(kmeans.inertia_)

plt.plot(K, sumsqdis)
plt.xlabel("Values of K") 
plt.ylabel("Sum of squared distances/Inertia") 
plt.title("Elbow Method For Optimal k")
plt.show()
```

```python
# By looking above data, it looks like 4 will be optimal number of group (cluster)
```

```python
#Doing K mean clustering, Trying to create 4 cluster
```

```python
kmeans = KMeans(n_clusters=6, max_iter=50,random_state=0)
kmeans.fit(reduced_rep_batting_data)
```

```python
pca_data=pd.DataFrame(reduced_rep_batting_data)
pca_data["cluster_kmeans"] = kmeans.labels_ 
display(pca_data)
```

```python
# Assign the label to the output column in the original players
data["cluster_kmeans"] = kmeans.labels_ 
data["pc1"]=reduced_rep_batting_data[:,0]
data["pc2"]=reduced_rep_batting_data[:,1]
#isplay(data[data["cluster_kmeans"] ==5])
#rint(display(data[data["Player"] == "AB de Vlliers"]))

columns = data.columns.tolist()
print(columns)
display(data)
```

```python
#Display the data
for i in range(0,6):
    print("Cluster Number ", i)
    temp=data[data["cluster_kmeans"] == i]
    display(temp[['Player', 'Runs', 'Average', 'SR', 'cluster_kmeans']])
```

```python
#Getting last name Just for display
data.loc[data['Player'].str.split().str.len() >=2, 'lname'] = data['Player'].str.split().str[-1]
display(data)
```

```python
sns.set(style="white")
plt.figure(figsize=(16,8))
ax = sns.lmplot(x="pc1", y="pc2", hue='cluster_kmeans', data = data, legend=False,
fit_reg=False, size = 15, scatter_kws={"s": 100})
texts = []
for x, y, s in zip(data.pc1, data.pc2, data.lname):
    texts.append(plt.text(x, y, s))
#ax.set(ylim=(2, 2))
plt.tick_params(labelsize=12)
plt.title("KMeans Clustering - Top 6 Batsmen Group",fontsize = 15 ) #title
plt.xlabel("PC 1", fontsize = 10)
plt.ylabel("PC 2", fontsize = 10)
plt.legend()
plt.show()
```

```python
fig = px.sunburst(data, path=['cluster_kmeans'],color='cluster_kmeans')
fig.show()
```
